```python
#Section 2 Data Preparation
    #Handle nulls.
	#Scale variables as necessary.
	#Encode variables as necessary.
	#Drop features you don't want.
	#Generate additional features if you want.
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
from sklearn.cluster import KMeans
from sklearn.preprocessing import OneHotEncoder

bank_full.dropna() #dropping rows that contain NAs

encoder = OneHotEncoder(sparse_output=False).set_output(transform='pandas')

bank_ohe = encoder.fit_transform(bank_full[['education']])
#I did my encoding on education to see if the level of education impacts getting a loan or not.

bank_combined = pd.concat([bank_full, bank_ohe], axis=1).drop(columns = ['education'])
#combined data back to the main dataframe and droped the main education column since I've encoded it. 
bank_combined
```

```python

titles = list(bank_combined.columns)
titles

order = ['age', 'job', 'marital', 'default','housing','loan','contact','month','day_of_week','duration','campaign','pdays','previous','poutcome','emp.var.rate','cons.price.idx','cons.conf.idx','euribor3m','nr.employed','education_basic.4y','education_basic.6y','education_basic.9y','education_high.school','education_illiterate','education_professional.course', 'education_university.degree', 'education_unknown','y',]
bank_combined = bank_combined[order]
bank_combined

bank_combined.dtypes
#reording column "y" to the end because it was placed in the middle of my integer data
```
